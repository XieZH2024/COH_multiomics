```python
import scanpy as sc
import anndata as ad
import matplotlib.pyplot as plt
```

```python
adata = sc.read_h5ad('data/snRNA_data/all_celltype.h5ad')
```

```python
sc.tl.umap(adata,min_dist=0.6,spread=1)
```

```python
adata.uns['celltype_colors'] = ['#4b95b7','#bb2232','#4c9942','#bee097','#c7b3d3','#ef9ba7','#9a6ab0','#df8139','#b5d0d9']
```

```python
with plt.rc_context({"figure.figsize": (5, 5)}):
    sc.pl.umap(adata, color='celltype', frameon=False, size=1, alpha=1)
```

```python
adata.obs['group'] = 'COH'
adata.obs.loc[adata.obs['condition'].isin(['Contralateral']),'group'] = 'Contralateral'
adata.obs['group'] = adata.obs['group'].astype('category')
adata.obs['group'] = adata.obs['group'].cat.reorder_categories(['Contralateral', 'COH'])
adata.obs['group'].value_counts()
```

```python
adata.uns['group_colors'] = ['#58585B','#4F8EE6']
```

```python
with plt.rc_context({"figure.figsize": (5, 5)}):
    sc.pl.umap(adata, color='group', frameon=False, size=1.2, alpha=1, sort_order=True)
```
