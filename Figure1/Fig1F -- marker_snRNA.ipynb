```python
markers = ['Pax6','Nrxn2','Gfap','Pdgfra','Grik1','Vsx2','Arr3','Opn1mw','Onecut1','Lhx1',
           'Hexb','Cx3cr1','Slc1a3','Aqp4','Rbpms','Slc17a6','Rho','Nrl']
```

```python
my_reds_rev = mcolors.LinearSegmentedColormap.from_list(
    "WhiteToLightRed", 
    ["#FBF0EB", "#f24d1f"]
)
```

```python
with plt.rc_context({"figure.figsize": (8, 4)}):
    sc.pl.dotplot(adata,markers,"celltype",swap_axes=True,standard_scale='var',cmap=my_reds_rev, linewidth=0)
```
