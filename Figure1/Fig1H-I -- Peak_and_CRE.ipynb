```r
library(ggplot2)
library(dplyr)
library(tidyr)
```

```r
df <- read.csv("data/snATAC_data/Peak_and_CRE/all_Peak_freq.csv", stringsAsFactors = FALSE)
df <- df[df$celltype != 'Union',]
total <- aggregate(freq ~ celltype, data = df, sum)
total <- total[order(total$freq, decreasing = FALSE), ]
celltype_order <- total$celltype
```

```r
total$celltype <- factor(total$celltype, levels = celltype_order)
ggplot(total, aes(x = celltype, y = freq)) +
  geom_col() +
  coord_flip() +
  labs(x = NULL, y = "total freq") +
  theme_classic()
```

```r
df$celltype <- factor(df$celltype, levels = celltype_order)
ggplot(df, aes(x = celltype, y = freq, fill = peaktype)) +
  geom_col(position = "fill") +
  coord_flip() +
  labs(x = NULL, y = "Proportion", fill = "peaktype") +
  theme_classic()
```
