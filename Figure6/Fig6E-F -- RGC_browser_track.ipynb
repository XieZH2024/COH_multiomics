```r
library(ArchR)
addArchRGenome("mm10")
library(BSgenome.Mmusculus.UCSC.mm10)
```

```r
addArchRThreads(threads = 30)
```

```r
rgc <- loadArchRProject(path = "data/snATAC_data/rgc")
```

```r
p <- plotBrowserTrack(
    ArchRProj = rgc, 
    groupBy = "rgc_sub", 
    geneSymbol = 'Mtor', 
    upstream = 150000,
    downstream = 70000,
    loops = getPeak2GeneLinks(rgc),ylim=c(0,1)
)
```

```r
grid::grid.newpage()
grid::grid.draw(p$Mtor)
```
