```r
library(Seurat)
library(ggplot2)
library(ggpubr)
```

```r
seu <- readRDS('data/snRNA_data/RGC.rds')
```

```r
options(repr.plot.height = 3, repr.plot.width = 5.5)
VlnPlot(object = seu, features = 'Atf4', group.by = "condition", pt.size = 0, y.max = 3) +
  labs(title = "Atf4", y = "Expression Level",x = "") + theme(legend.position = "none") 
```
