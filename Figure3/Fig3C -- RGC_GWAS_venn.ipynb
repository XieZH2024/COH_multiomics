```r
library(VennDiagram)
library(svglite)
library(dplyr)
```

```r
mmu_df <- read.csv('data/Venn/gwas_mouse_genes.csv')
mmu <- mmu_df$mmusculus_gene_name
```

```r
deg <- read.csv('data/snRNA_data/DEG/Retinal ganglion cell.csv')
```

```r
deg_2w <- deg[deg$COH_2w_pvals_adj<0.01 & abs(deg$COH_2w_logfoldchanges)>=0.25,'COH_2w_names']
deg_4w <- deg[deg$COH_4w_pvals_adj<0.01 & abs(deg$COH_4w_logfoldchanges)>=0.25,'COH_4w_names']
deg_8w <- deg[deg$COH_8w_pvals_adj<0.01 & abs(deg$COH_8w_logfoldchanges)>=0.25,'COH_8w_names']
```

```r
venn_list <- list(GWAS = mmu, COH_8w = deg_8w, COH_2w = deg_2w, COH_4w = deg_4w)
```

```r
venn.plot <- venn.diagram(x = venn_list, filename = 'venn.tiff',disable.logging = TRUE,
    fill = c("#6BAED6", "#FD8D3C", "#74C476", "#E377C2"), alpha = 0.4)
```

```r
mmu_df <- mmu_df %>% mutate(if_2w_DEG=ifelse(mmu_df$mmusculus_gene_name %in% deg_2w,'Yes','No')) %>%
                     mutate(if_4w_DEG=ifelse(mmu_df$mmusculus_gene_name %in% deg_4w,'Yes','No')) %>%
                     mutate(if_8w_DEG=ifelse(mmu_df$mmusculus_gene_name %in% deg_8w,'Yes','No'))
```

```r
write.csv(mmu_df,'data/Venn/gwas_mouse_genes.csv',row.names=FALSE)
```
