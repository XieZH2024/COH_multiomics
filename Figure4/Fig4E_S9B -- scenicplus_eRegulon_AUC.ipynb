# AUC z-score

```python
erg = metadata_filtered['Gene_signature_name'].unique()
erg
```

```python
adata = scplus_mdata['direct_gene_based_AUC'].copy()
adata = adata[:,adata.var_names.isin(erg)]
adata.obs['condition'] = scplus_mdata['scRNA_counts'].obs['condition'].tolist()
```

```python
auc = pd.DataFrame(adata.X, index=adata.obs_names, columns=adata.var_names)
```

```python
from scipy.stats import zscore
auc_z = auc.apply(zscore, axis=0)
```

```python
auc_z["condition"] = adata.obs["condition"].values
```

```python
mean_z = auc_z.groupby("condition").mean()
```

```python
mean_A = mean_z.loc["Contra"]
mean_B = mean_z.loc["COH_2w"]
mean_C = mean_z.loc["COH_4w"]
mean_D = mean_z.loc["COH_8w"]
```

```python
from scipy.stats import ttest_ind
from statsmodels.stats.multitest import multipletests
```

```python
expressed = ((mean_A > 0) | (mean_B > 0) & (abs(mean_B - mean_A)>0.1))
regs = expressed[expressed].index
pvals = []
for reg in mean_A.index:
    if reg in regs:
        A = auc_z.loc[auc_z["condition"]=="Contra",reg]
        B = auc_z.loc[auc_z["condition"]=="COH_2w",reg]
        p = ttest_ind(A, B, equal_var=True).pvalue
        pvals.append(p)
    else:
        pvals.append(1)
padj_2w = multipletests(pvals, method="fdr_bh")[1]
diff_2w = mean_B - mean_A
```

```python
expressed = ((mean_A > 0) | (mean_C > 0) & (abs(mean_C - mean_A)>0.1))
regs = expressed[expressed].index
pvals = []
for reg in mean_A.index:
    if reg in regs:
        A = auc_z.loc[auc_z["condition"]=="Contra",reg]
        C = auc_z.loc[auc_z["condition"]=="COH_4w",reg]
        p = ttest_ind(A, C, equal_var=True).pvalue
        pvals.append(p)
    else:
        pvals.append(1)
padj_4w = multipletests(pvals, method="fdr_bh")[1]
diff_4w = mean_C - mean_A
```

```python
expressed = ((mean_A > 0) | (mean_D > 0) & (abs(mean_D - mean_A)>0.1))
regs = expressed[expressed].index
pvals = []
for reg in mean_A.index:
    if reg in regs:
        A = auc_z.loc[auc_z["condition"]=="Contra",reg]
        D = auc_z.loc[auc_z["condition"]=="COH_8w",reg]
        p = ttest_ind(A, D, equal_var=True).pvalue
        pvals.append(p)
    else:
        pvals.append(1)
padj_8w = multipletests(pvals, method="fdr_bh")[1]
diff_8w = mean_D - mean_A
```

```python
result = pd.DataFrame({
    "diff_2w": diff_2w,
    "diff_4w": diff_4w,
    "diff_8w": diff_8w,
    "padj_2w": padj_2w,
    "padj_4w": padj_4w,
    "padj_8w": padj_8w
})
```

```python
result.to_csv('data/scenicplus/AUC_test.csv')
```
