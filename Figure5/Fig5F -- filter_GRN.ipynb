```r
library(tidyverse)
library(stringr)
```

```r
enrich <- read.csv('data/scenicplus/result_deduplicated.csv')
```

```r
filtered_enrich <- enrich %>%
  filter(str_detect(Description, 
         regex("calcium|hippo|endoplasmic reticulum|apoptosis|apoptotic|autophagy|neurodegeneration", ignore_case = TRUE)))
```

```r
unique_genes <- filtered_enrich %>%
  pull(geneID) %>%
  str_split("/") %>%
  unlist() %>%
  str_trim() %>%
  unique() %>%
  sort()
length(unique_genes)
```

```r
sp_result <- read.csv('data/scenicplus/eRegulon_filtered.csv')
```

```r
tf <- unique(sp_result$TF)
```

```r
filtered_sp_result <- sp_result[(sp_result$Gene %in% unique_genes)&(sp_result$TF %in% unique_genes),]
```

```r
write.csv(filtered_sp_result,'data/scenicplus/scenic_plus_filtered.csv',row.names=FALSE)
```
